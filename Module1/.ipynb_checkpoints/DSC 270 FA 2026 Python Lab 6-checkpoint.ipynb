### 2.1

**Supposed to:** count how many temperatures in the list are below freezing (32). The answer for this list is 3.

The code above has a syntax error missing a semicolon at the end of the for loop.

```python
temps = [28, 35, 31, 40, 30]
freezing = 0
for t in temps:
    if t < 32:
        freezing = freezing + 1
print(freezing)
```

### 2.2

**Supposed to:** add up a list of minutes. The values are strings because they were read from a file. The answer is 135.

This is an example of a runtime error. The type string is added to type int which results in a TypeError

```python
minutes = ["45", "30", "60"]
total = 0
for m in minutes:
    total = total + int(m)
print(total)
```

### 2.3

**Supposed to:** find the shortest checkout, in days. The answer for this list is 7.

This is an example of a logical error which assumes that 0 is the shortest checkout by default. If the list contains values all above zero while still having the smallest value, the program will still print the initial 0 that we defaulted to.

```python
days = [14, 21, 7, 30]
shortest = days[0]
for d in days:
    if d < shortest:
        shortest = d
print(shortest)
```

### 2.5

**Supposed to:** read `checkouts.csv`, print the column names as a list of three clean strings (no quotation marks), then count how many checkouts each branch has and print that dictionary.

Before you decide anything, print the first two lines of the file and look at them.

The code assumes that the data is comma separated when infact, the data is semi-colon separated, and the titles need a replace method to get rid of the quotation marks.

```python
lines = open("checkouts.csv").read().strip().replace('"',"").split("\n")

headings = lines[0].split(";")
print(headings)

counts = {}
for line in lines[1:]:
    branch = line.split(";")[0]
    counts[branch] = counts.get(branch, 0) + 1
print(counts)
```
