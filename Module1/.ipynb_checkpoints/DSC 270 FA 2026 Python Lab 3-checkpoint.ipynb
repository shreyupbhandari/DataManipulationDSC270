## Task 5: The Cleaning Loop

Time to fix the mess. In a new code cell below this instruction, use the transform-and-collect pattern to build a new list named `clean_headings`:

- Start with an empty list.
- Loop over `raw_headings`.
- For each heading: remove the whitespace at the ends, convert to lowercase, and replace each space **inside** the name with an underscore (`_`). Then append the result.

Print `clean_headings` when you are done. The first cleaned heading should be `order_id`. Leave `raw_headings` unchanged.

```python
cleanedList=[]
for headings in raw_headings:
    cleanedList.append(headings.strip().lower().replace(" ","_"))
print(cleanedList)
```

## Task 6: Normalize and Count

The payment column is just as messy. Run the next cell to load it.

```python
# Run this cell first
payments = ["card", " Cash", "CARD ", "cash", "Card", "  card", "cash ", "CASH", "card", " card"]
```

In a new code cell below this instruction, count how many customers paid by card. Loop over the list, normalize each value before comparing it to `"card"`, and keep a running count.

Then use an f-string to print a one-sentence report giving the number of card payments and the percentage of all payments they represent, with the percentage rounded to one decimal place.

```python
cleanedPaymentList=[]
for rawPayments in payments:
    cleanedPaymentList.append(rawPayments.strip().lower().replace(" ","_"))
cardPaymentCount=0
for payment in cleanedPaymentList:
    if payment=="card":
        cardPaymentCount+=1
fractionOfCardPayments=(cardPaymentCount/len(cleanedPaymentList))
print(f"The number of card payments are {cardPaymentCount} and it is {fractionOfCardPayments*100:.1f} percent of the total payment methods.")
```
