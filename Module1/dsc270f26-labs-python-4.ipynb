## Task 5: One Row Becomes a Record

In a new code cell, take `rows[0]`, split it on commas, and build a dictionary named `first` with exactly these five keys: `order_id`, `item_name`, `price`, `qty`, and `payment_method`.

Watch the types, because everything arrives as text:

- `order_id` and `payment_method` stay strings, with the whitespace stripped off.
- `item_name` stays a string, stripped **and** lowercased.
- `price` has to end up a `float`, which means the dollar sign has to go first.
- `qty` has to end up an `int`.

Print `first`. Then, on a second line, print what that order was worth in total (the price times the quantity) using an f-string with two decimal places.

````python
values = rows[0].split(",")
first={"order_id":values[0].strip(),
       "item_name":values[1].strip().lower(), 
       "price":float(values[2].strip().replace("$","")), 
       "qty":int(values[3].strip()),
       "payment_method":values[4].strip()}
print(first)
````

## Task 6: Build the Whole Table

In a new code cell, do to every row what you just did to `rows[0]`, collecting the records into a list named `table`.

1. Print how many records are in `table`.
2. Print `table[3]`.
3. Print the item name of the **last** order, using a negative index and a key.

````python
table=[]
for row in rows:
    value=row.split(",")
    record={"order_id":value[0].strip(),
       "item_name":value[1].strip().lower(), 
       "price":float(value[2].strip().replace("$","")), 
       "qty":int(value[3].strip()),
       "payment_method":value[4].strip()
           }
    table.append(record)
print(table[-1])
````
