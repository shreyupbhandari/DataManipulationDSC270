## Task 2: Write `tidy` and Use It on a List

Run the next cell. It is a list of city names typed in by several different people.

```python
# Run this cell first
cities = [" danville", "LEXINGTON ", "louisville", " Berea", "frankfort  ", "Richmond"]
```

In a new code cell:

1. Define a function `tidy(s)` that takes one string and **returns** it with the whitespace removed from both ends and the first letter of each word capitalized. (The string method `.title()` does the capitalizing.) Use `return`, not `print`.
2. Test it by calling `tidy(" danville")` and `tidy("LEXINGTON ")` and printing the results. You should see `Danville` and `Lexington`.
3. Use a `for` loop to build a new list named `cleaned` by calling `tidy` on each item of `cities`. Print `cleaned`.

Then run the cell below. It builds the same list with a list comprehension. You do not have to write one, only read this one. The last line should print `True`.

```python
def tidy(s):
    return s.strip().title()
tidy("Danville")
tidy("Lexington")

cleaned=[]
for city in cities:
    cleaned.append(tidy(city))
print(cleaned)
```

```python
# Run this cell after you have defined tidy and built cleaned.
# Read it from the "for": for each c in cities, collect tidy(c).
one_liner = [tidy(c) for c in cities]
print(one_liner)
print(one_liner == cleaned)
```

## Task 3: A Dictionary In, Two Values Out

Run the next cell. It holds two dictionaries of counts.

```python
# Run this cell first
states = {"KY": 14, "OH": 4, "TN": 3, "IN": 2, "IL": 1}
languages = {"Python": 6, "R": 3, "Excel": 12, "SQL": 2, "none": 1}
```

In a new code cell:

1. Define a function `biggest(counts)` that takes a dictionary of counts and returns **two** values: the key with the largest count, and that count. Loop over the keys, keeping track of the best key and the best count seen so far. Return both on one line, separated by a comma.
2. Call `biggest(states)` and store what comes back in a variable named `result`. Print `result` and print `type(result)`.
3. Call `biggest(states)` again, this time unpacking the two values into variables named `state` and `n`. Use an f-string to print a sentence such as `Most students are from KY (14).`
4. Do the same with `languages`, unpacking into `language` and `n`, and print a similar sentence. The function should work without any changes.

```python
def biggest(counts):
    count=0
    bestKey=""
    for key,values in counts.items():
        if values>count:
            count=values
            bestKey=key
    return bestKey,count
result = biggest(states)
print(result)
print(type(result))
state, n=biggest(states)
print(f"Most students are from {state}({n}).")
language, n=biggest(languages)
print(f"Most languages are {language}({n}).")
```
