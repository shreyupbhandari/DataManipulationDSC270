## Task 1: Calculate Inventory Distribution

You are managing a warehouse for a retail company and receive a shipment of 487 pairs of shoes to be distributed among 25 stores.

In a new code cell below this instruction, determine:

1. The number of pairs of shoes each store will receive.
2. The number of pairs of shoes that will remain after distribution.

Use appropriate Python operators to solve this problem.

## Task 3: Calculate Satellite Orbital Period

The orbital period of a satellite around a planet can be calculated using the following formula derived from Kepler's Third Law:

$$ T = 2\pi \sqrt{\frac{a^3}{\mu}} $$

Where:

- $ T $ is the orbital period in seconds.
- $ a $ is the semi-major axis of the orbit in meters.
- $ \mu $ (Greek letter mu) is the standard gravitational parameter of the planet in $ m^3/s^2 $.

For Earth, $ \mu $ is approximately $ 3.986 \times 10^{14} \, m^3/s^2 $.

In a new code cell below this instruction, calculate and print the orbital period for a satellite that has a semi-major axis of 42,164 km. **Convert this value to meters before using it in the formula.**

To compute the square root here, you can use the `**` operator to raise a number to the power of 0.5, like this: `x ** 0.5`.

```python
import math
semiMajorAxis=42164*1000
gravitationalParameter=3.986e14

orbitalPeriod=2*math.pi*(semiMajorAxis**3/gravitationalParameter)**(1/2)
print(f'The orbital period for satellite that has a semi-major axis of {semiMajorAxis} around the Earth is: {orbitalPeriod}')
```

## Task 5: Implement the ReLU Function
